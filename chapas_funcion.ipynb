````python


# ---------------------------------------------------------------------------
# 1. AUTENTICACIÓN E INICIALIZACIÓN
# ---------------------------------------------------------------------------
def inicializar_cliente_gemini(project_id: str = None, location: str = "us-central1") -> genai.Client:
    """Inicializa y retorna el cliente oficial de Google GenAI."""
    auth.authenticate_user()
    if project_id:
        client = genai.Client(vertexai=True, project=project_id, location=location)
    else:
        client = genai.Client()
    print("✅ Cliente de Gemini inicializado con éxito.")
    return client

def calcular_costo(input_tokens: int, output_tokens: int, cached_tokens: int = 0, model_name: str = "gemini-2.5-flash") -> float:
    """Calcula el costo aproximado en USD basado en los precios de Gemini 2.5 Flash."""
    # Tarifas estimadas por millón de tokens (Gemini 2.5 Flash)
    PRECIO_INPUT = 0.075 / 1_000_000
    PRECIO_CACHED = 0.01875 / 1_000_000
    PRECIO_OUTPUT = 0.30 / 1_000_000

    costo = (input_tokens * PRECIO_INPUT) + (cached_tokens * PRECIO_CACHED) + (output_tokens * PRECIO_OUTPUT)
    return costo


````

````python
def crear_context_cache(
    client: genai.Client, texto_contexto_estatico: str, ttl_segundos: int = 7200
) -> str:
  """Sube el contexto estático a Google Cloud usando Context Caching.

  Utiliza tipos directos e intenta fallback por diccionario si la
  serialización del SDK falla.
  """
  if not texto_contexto_estatico or not texto_contexto_estatico.strip():
    raise ValueError("❌ 'prompt_cache' está vacío.")

  # 1. Validar tokens reales
  conteo = client.models.count_tokens(
      model="gemini-2.5-flash", contents=texto_contexto_estatico
  )
  tokens_totales = conteo.total_tokens
  print(f"📊 Tokens detectados en el contexto estático: {tokens_totales}")

  if tokens_totales < 1024:
    raise ValueError(
        f"❌ El contexto solo tiene {tokens_totales} tokens. Mínimo requerido:"
        " 1024."
    )

  print("⏳ Creando Context Cache en Google Cloud...")

  # 2. Intento 1: Usando types.Part(text=...) sin from_text
  try:
    contenido_objetos = [
        types.Content(
            role="user", parts=[types.Part(text=texto_contexto_estatico)]
        )
    ]
    cache = client.caches.create(
        model="gemini-2.5-flash",
        config=types.CreateCachedContentConfig(
            contents=contenido_objetos, ttl=f"{ttl_segundos}s"
        ),
    )
  except Exception as e:
    # Intento 2 (Fallback): Diccionario puro sin envoltorios de clases
    print("⚠️ Reintentando con estructura dict para evitar serialización del SDK...")
    contenido_dict = [
        {"role": "user", "parts": [{"text": texto_contexto_estatico}]}
    ]
    cache = client.caches.create(
        model="gemini-2.5-flash",
        config=types.CreateCachedContentConfig(
            contents=contenido_dict, ttl=f"{ttl_segundos}s"
        ),
    )

  print(f"✅ Cache creado exitosamente con ID: {cache.name}")
  return cache.name
````

````python
def procesar_placa_2_pasadas(
    client: genai.Client,
    imagen: str | Path | Image.Image | Any,
    id_cache: str,
    prompt_TV_1: str,
    prompt_TV_2: str,
    model: str = "gemini-2.5-flash",
    thinking_budget: int = 1024,
) -> str:
  """Ejecuta el análisis de la chapa en 2 pasadas usando Context Caching

  y motor de razonamiento (Thinking Mode) para Gemini 2.5 Flash.

  Acepta tanto una ruta de archivo (str/Path) como una imagen ya cargada (PIL
  Image). Retorna únicamente el string con el JSON final completado.
  """
  # Si se pasa una ruta de texto o Path, la cargamos automáticamente con PIL
  if isinstance(imagen, (str, Path)):
    imagen_chapa = Image.open(imagen)
  else:
    imagen_chapa = imagen

  # --- PASADA 1: Transcripción inicial y detección de dudas ---
  respuesta_pasada_1 = client.models.generate_content(
      model=model,
      contents=[imagen_chapa, prompt_TV_1],
      config=types.GenerateContentConfig(
          cached_content=id_cache, response_mime_type="application/json"
      ),
  )

  json_semi_lleno = respuesta_pasada_1.text

  # --- PASADA 2: Deducción según normas IEC y resolución de ambigüedades ---
  prompt_pasada_2_formatted = prompt_TV_2.format(json_pasada_1=json_semi_lleno)

  respuesta_final = client.models.generate_content(
      model=model,
      contents=[imagen_chapa, prompt_pasada_2_formatted],
      config=types.GenerateContentConfig(
          cached_content=id_cache,
          thinking_config=types.ThinkingConfig(
              thinking_budget=thinking_budget
          ),
          response_mime_type="application/json",
      ),
  )

  return respuesta_final.text
````

````python
# Silenciar warnings
logging.getLogger("google_genai.models").setLevel(logging.ERROR)

# ---------------------------------------------------------------------------
# FUNCIONES AUXILIARES Y DE REINTENTO
# ---------------------------------------------------------------------------
def es_error_de_cuota(excepcion):
    """Retorna True si el error es por saturación (429/503)."""
    error_str = str(excepcion).lower()
    return "429" in error_str or "resource exhausted" in error_str or "503" in error_str

def limpiar_json(texto):
    """Extrae el JSON válido ignorando las etiquetas Markdown (```json) de Gemini."""
    match = re.search(r'```json\n(.*?)\n```', texto, re.DOTALL)
    if match:
        texto = match.group(1)
    return json.loads(texto)

def aplanar_diccionario(d, parent_key='', sep='_'):
    """Aplana un JSON anidado. Ejemplo: metadatos: {año: 2020} -> metadatos_año: 2020"""
    items = []
    for k, v in d.items():
        new_key = f"{parent_key}{sep}{k}" if parent_key else k
        if isinstance(v, dict):
            items.extend(aplanar_diccionario(v, new_key, sep=sep).items())
        elif isinstance(v, list):
            items.append((new_key, ", ".join(map(str, v))))
        else:
            items.append((new_key, v))
    return dict(items)

def conectar_google_sheet(nombre_sheet):
    """Autentica en Colab y abre la Planilla."""
    auth.authenticate_user()
    creds, _ = default()
    gc = gspread.authorize(creds)

    if "[docs.google.com/spreadsheets](https://docs.google.com/spreadsheets)" in nombre_sheet:
        return gc.open_by_url(nombre_sheet)
    return gc.open_by_key(nombre_sheet) if "/" not in nombre_sheet and len(nombre_sheet) > 30 else gc.open(nombre_sheet)

# ---------------------------------------------------------------------------
# FUNCIÓN DE LLAMADO A GEMINI PROTEGIDA
# ---------------------------------------------------------------------------
@retry(
    wait=wait_exponential(multiplier=2, min=10, max=120),
    stop=stop_after_attempt(8),
    retry=retry_if_exception(es_error_de_cuota),
    reraise=True
)



def obtener_mime_type(ruta_archivo):
  mime, _ = mimetypes.guess_type(ruta_archivo)
  if not mime:
    ext = os.path.splitext(ruta_archivo)[1].lower()
    if ext in ['.jpg', '.jpeg']:
      return 'image/jpeg'
    elif ext == '.png':
      return 'image/png'
    elif ext == '.pdf':
      return 'application/pdf'
    return 'image/jpeg'
  return mime


def procesar_chapa_con_reintentos(
    client, ruta_archivo, id_cache, prompt_TV_1, prompt_TV_2
):
  mime_type = obtener_mime_type(ruta_archivo)

  # Leer el archivo local en bytes
  with open(ruta_archivo, 'rb') as f:
    bytes_imagen = f.read()

  # Crear la parte de contenido multimedia desde bytes
  parte_media = types.Part.from_bytes(data=bytes_imagen, mime_type=mime_type)

  # Pasada 1
  resp_1 = client.models.generate_content(
      model='gemini-2.5-flash',
      contents=[parte_media, prompt_TV_1],
      config=types.GenerateContentConfig(
          cached_content=id_cache, response_mime_type='application/json'
      ),
  )

  # Pasada 2 (Refinamiento/Consolidación)
  resp_2 = client.models.generate_content(
      model='gemini-2.5-flash',
      contents=[parte_media, resp_1.text, prompt_TV_2],
      config=types.GenerateContentConfig(
          cached_content=id_cache, response_mime_type='application/json'
      ),
  )

  return resp_2.text

# ---------------------------------------------------------------------------
# BUCLE PRINCIPAL DE PROCESAMIENTO
# ---------------------------------------------------------------------------



def procesar_carpeta_a_sheets(
    client,
    id_cache,
    carpeta_origen,
    url_sheet,
    nombre_pestaña,
    prompt_TV_1,
    prompt_TV_2,
):
  print('🔗 Conectando a Google Sheets...')
  sheet = conectar_google_sheet(url_sheet).worksheet(nombre_pestaña)
  encabezados = sheet.row_values(1)

  # ---------------------------------------------------------------------------
  # NUEVO: IDENTIFICAR ARCHIVOS YA PROCESADOS EN LA PLANILLA
  # ---------------------------------------------------------------------------
  try:
    # Busca en qué número de columna está "nombre_imagen" (1-based para Sheets)
    col_idx = encabezados.index("nombre_imagen") + 1
    # Trae todos los valores de esa columna omitiendo el encabezado
    nombres_procesados = set(sheet.col_values(col_idx)[1:])
    print(f"✅ Se detectaron {len(nombres_procesados)} registros ya guardados en Sheets.")
  except ValueError:
    print("⚠️ No se encontró la columna 'nombre_imagen' en Sheets. Se procesará todo desde cero.")
    nombres_procesados = set()

  # ---------------------------------------------------------------------------
  # BÚSQUEDA RECURSIVA EN SUBCARPETAS
  # ---------------------------------------------------------------------------
  archivos_totales = []
  extensiones_validas = {
      '.pdf', '.jpg', '.jpeg', '.png', '.bmp', '.webp', '.heic'
  }

  print(f'🔍 Escaneando archivos en: {carpeta_origen} ...')

  for raiz, _, ficheros in os.walk(carpeta_origen, followlinks=True):
    for f in ficheros:
      if not f.startswith('.') and not f.startswith('~'):
        ext = os.path.splitext(f)[1].lower()
        if ext in extensiones_validas:
          archivos_totales.append(os.path.join(raiz, f))

  # Filtrar los que ya están en la planilla
  archivos_pendientes = [
      ruta for ruta in archivos_totales
      if os.path.basename(ruta) not in nombres_procesados
  ]

  archivos_pendientes = sorted(archivos_pendientes)
  total_pendiente = len(archivos_pendientes)

  if total_pendiente == 0:
    print(f'🎉 No hay nada nuevo por hacer. Los {len(archivos_totales)} archivos ya están procesados.')
    return

  print(f'📂 ¡Quedan {total_pendiente} archivos nuevos/pendientes para procesar!\n')

  # ---------------------------------------------------------------------------
  # PROCESAMIENTO UNO A UNO (SOLO PENDIENTES)
  # ---------------------------------------------------------------------------
  for idx, ruta in enumerate(archivos_pendientes, start=1):
    nombre_archivo = os.path.basename(ruta)
    print(
        f'[{idx}/{total_pendiente}] Procesando: {nombre_archivo} ... ', end='', flush=True
    )

    try:
      resultado_texto = procesar_chapa_con_reintentos(
          client, ruta, id_cache, prompt_TV_1, prompt_TV_2
      )

      # Parsear JSON y agregar nombre de archivo
      datos_planos = json.loads(resultado_texto)
      datos_planos['nombre_imagen'] = nombre_archivo

      # Mapear datos a las columnas de la fila 1
      fila_a_insertar = [datos_planos.get(col, '') for col in encabezados]

      # Insertar en Sheets
      sheet.append_row(fila_a_insertar, value_input_option='USER_ENTERED')
      print('✅ OK')

    except Exception as e:
      print(f'❌ Error: {e}')

    # Pacing de cuotas entre llamadas (10 segs)
    if idx < total_pendiente:
      time.sleep(10)

  print('\n🎉 Proceso completado exitosamente.')
````

````python
import pandas as pd
import re
import time
import gspread
from google.colab import auth
from google.auth import default

# ------------------------------------------------------------------------------
# 1. AUTENTICACIÓN Y CONEXIÓN A GOOGLE SHEETS
# ------------------------------------------------------------------------------
def conectar_google_sheets(nombre_o_id="TV"):
    auth.authenticate_user()
    creds, _ = default()
    gc = gspread.authorize(creds)
    return gc.open(nombre_o_id)

def ejecutar_con_reintento(func, *args, **kwargs):
    """Ejecuta una función de gspread. Si salta el error 429 (cuota), espera y reintenta."""
    max_intentos = 5
    espera = 10  # segundos
    for intento in range(max_intentos):
        try:
            return func(*args, **kwargs)
        except gspread.exceptions.APIError as e:
            if "429" in str(e):
                print(f"Límite de cuota alcanzado (429). Esperando {espera} segundos para reintentar (Intento {intento+1}/{max_intentos})...")
                time.sleep(espera)
                espera *= 2  # Aumenta el tiempo de espera exponencialmente
            else:
                raise e
    raise RuntimeError("Se superó el número máximo de reintentos por cuota de API.")

# ------------------------------------------------------------------------------
# 2. BÚSQUEDA ROBUSTA DE COLUMNAS
# ------------------------------------------------------------------------------
def buscar_columna(df, patrones):
    for patron in patrones:
        for col in df.columns:
            if str(col).strip().lower() == patron.lower():
                return df[col]
        for col in df.columns:
            if re.search(patron, str(col), re.IGNORECASE):
                return df[col]
    return pd.Series([''] * len(df), index=df.index)

# ------------------------------------------------------------------------------
# 3. REGLAS DE TRANSFORMACIÓN A FORMATO SAP
# ------------------------------------------------------------------------------
def parse_voltaje_primario(v_prim):
    if pd.isna(v_prim) or str(v_prim).strip() == "":
        return ""
    try:
        val_str_raw = str(v_prim).replace(',', '.').split('/')[0].strip()
        val = float(re.sub(r'[^\d.]', '', val_str_raw))
        if val >= 1000:
            val = val / 1000
        val_formatted = f"{val:.3f}".rstrip('0').rstrip('.')
        return f"{val_formatted}/V3"
    except Exception:
        return str(v_prim)

def parse_voltaje_secundario(v_sec):
    if pd.isna(v_sec) or str(v_sec).strip() == "":
        return ""
    try:
        val_str_raw = str(v_sec).replace(',', '.').split('/')[0].strip()
        val = float(re.sub(r'[^\d.]', '', val_str_raw))
        val_formatted = f"{val:.3f}".rstrip('0').rstrip('.')
        return f"{val_formatted}/V3"
    except Exception:
        return str(v_sec)

def armar_relacion_tv(row):
    col_prim = next((c for c in row.index if re.search(r'primari', str(c), re.IGNORECASE)), None)
    v_prim = row.get(col_prim) if col_prim else ""
    prim_str = parse_voltaje_primario(v_prim)

    sec_cols = [c for c in row.index if re.search(r'(vt_sec\d+.*tension|sec\d+.*tension)', str(c), re.IGNORECASE)]
    sec_cols.sort()

    sec_strs = []
    for col in sec_cols:
        val = row.get(col)
        if pd.notna(val) and str(val).strip() != "":
            sec_strs.append(parse_voltaje_secundario(val))

    if not prim_str and not sec_strs:
        return ""
    if prim_str and sec_strs:
        return f"{prim_str} - {'-'.join(sec_strs)}"
    return prim_str or '-'.join(sec_strs)

def armar_prestacion_tv(row):
    cols = [c for c in row.index if re.search(r'(vt_sec\d+.*(carga|potencia|va|sr)|sec\d+.*(carga|potencia|va|sr))', str(c), re.IGNORECASE) and not re.search(r'tension|clase', str(c), re.IGNORECASE)]
    cols.sort()

    valores = []
    for col in cols:
        val = row.get(col)
        if pd.notna(val) and str(val).strip() != "":
            try:
                val_num = float(str(val).replace(',', '.'))
                val_str = str(int(val_num)) if val_num.is_integer() else str(val_num).replace('.', ',')
            except ValueError:
                val_str = str(val).strip()
            valores.append(val_str)

    return "-".join(valores)

def armar_clase_tv(row):
    cols = [c for c in row.index if re.search(r'(vt_sec\d+.*clase|sec\d+.*clase)', str(c), re.IGNORECASE)]
    cols.sort()

    valores = []
    for col in cols:
        val = row.get(col)
        if pd.notna(val) and str(val).strip() != "":
            val_str = str(val).strip().replace('.', ',')
            valores.append(val_str)

    return "-".join(valores)

def formatear_anio(val):
    if pd.isna(val) or str(val).strip() == "":
        return ""
    try:
        anio_int = int(float(str(val).replace('.', '').replace(',', '')))
        if 1900 <= anio_int <= 2050:
            return f"{anio_int // 1000}.{anio_int % 1000:03d}"
        return str(anio_int)
    except Exception:
        return str(val)

# ------------------------------------------------------------------------------
# 4. PROCESAMIENTO Y ESCRITURA CON CONTROL DE CUOTA
# ------------------------------------------------------------------------------
def procesar_tv_gsheet(nombre_archivo="TV"):
    print(f"Conectando a Google Sheets ({nombre_archivo})...")
    sh = ejecutar_con_reintento(conectar_google_sheets, nombre_archivo)

    wks1 = ejecutar_con_reintento(sh.worksheet, 'Hoja 1')
    datos_origen = ejecutar_con_reintento(wks1.get_all_records)
    df_origen = pd.DataFrame(datos_origen)

    if df_origen.empty:
        print("La 'Hoja 1' está vacía.")
        return

    df_sap = pd.DataFrame(index=df_origen.index)

    df_sap['MARCA'] = buscar_columna(df_origen, ['general_fabricante', 'fabricante', 'marca']).fillna('').astype(str).str.upper()
    df_sap['MODELO'] = buscar_columna(df_origen, ['general_modelo_tipo', 'modelo', 'tipo']).fillna('').astype(str)
    df_sap['NUMERO DE SERIE'] = buscar_columna(df_origen, ['general_numero_serie', 'numero_serie', 'serie', 'sn']).fillna('').astype(str)
    df_sap['RELACION (TV)'] = df_origen.apply(armar_relacion_tv, axis=1)
    df_sap['PRESTACIÓN (TV)'] = df_origen.apply(armar_prestacion_tv, axis=1)
    df_sap['CLASE (TV)'] = df_origen.apply(armar_clase_tv, axis=1)
    df_sap['AÑO FABRICACION'] = buscar_columna(df_origen, ['general_anio_fabricacion', 'anio_fabricacion', 'anio', 'año']).apply(formatear_anio)
    df_sap['Aislación (SF6/ACEITE/RESINA)'] = buscar_columna(df_origen, ['aislamiento_fluido', 'fluido', 'aislamiento']).fillna('').astype(str).str.upper()

    try:
        wks2 = ejecutar_con_reintento(sh.worksheet, 'Hoja 2')
        ejecutar_con_reintento(wks2.clear)
    except gspread.exceptions.WorksheetNotFound:
        wks2 = ejecutar_con_reintento(sh.add_worksheet, title='Hoja 2', rows=str(len(df_sap) + 10), cols=str(len(df_sap.columns) + 2))

    df_sap_limpio = df_sap.fillna('').astype(str)
    matriz_datos = [df_sap_limpio.columns.tolist()] + df_sap_limpio.values.tolist()

    # Pausa ligera de seguridad antes de escribir el bloque
    time.sleep(2)
    ejecutar_con_reintento(wks2.update, range_name='A1', values=matriz_datos)
    print("¡Proceso completado! Todos los campos mapeados se han escrito en 'Hoja 2'.")

if __name__ == '__main__':
    NOMBRE_DOCUMENTO = "TV"
    procesar_tv_gsheet(NOMBRE_DOCUMENTO)
````

````python
import re
import gspread
from google.colab import auth
import google.auth
from gspread_formatting import *

# 1. Autenticación
auth.authenticate_user()
creds, _ = google.auth.default()
gc = gspread.authorize(creds)

# 2. Configuración
NOMBRE_ARCHIVO = 'TV'
HOJA_DESTINO = 'Hoja 2'
HOJAS_REFERENCIA = ['Copia de TR', 'Copia de TB']

sh = gc.open(NOMBRE_ARCHIVO)
hoja2 = sh.worksheet(HOJA_DESTINO)

def normalizar(texto):
    """Limpia espacios extra, convierte a string y pasa a mayúsculas para comparar."""
    if texto is None:
        return ""
    txt = str(texto).strip().upper()
    # Reemplazar múltiples espacios por uno solo
    txt = re.sub(r'\s+', ' ', txt)
    return txt

def obtener_diccionario_ref(sheet):
    """Carga los datos de una hoja y crea una clave normalizada de Número de Serie."""
    datos = sheet.get_all_records()
    dict_ref = {}
    for fila in datos:
        # Buscar la clave del número de serie dinámicamente
        clave_sn = next((k for k in fila.keys() if 'SERIE' in str(k).upper() or 'SN' in str(k).upper()), None)
        if clave_sn:
            sn_val = normalizar(fila[clave_sn])
            if sn_val:
                # Normalizar todas las llaves y valores de la fila
                dict_ref[sn_val] = {normalizar(k): normalizar(v) for k, v in fila.items()}
    return dict_ref

# 3. Cargar referencias
db_referencia = {}
for nombre in HOJAS_REFERENCIA:
    try:
        ref_sheet = sh.worksheet(nombre)
        db_referencia.update(obtener_diccionario_ref(ref_sheet))
    except gspread.WorksheetNotFound:
        print(f"⚠️ Advertencia: No se encontró la hoja '{nombre}'.")

print(f"✅ Se cargaron {len(db_referencia)} registros de referencia desde SAP.\n")

# 4. Lectura de Hoja 2
datos_hoja2 = hoja2.get_all_values()
if not datos_hoja2:
    raise Exception("La 'Hoja 2' está vacía.")

encabezados_h2 = [normalizar(h) for h in datos_hoja2[0]]

# Estilos de Color
fmt_verde = cellFormat(backgroundColor=color(0.7, 0.9, 0.7))    # Todo coincide
fmt_amarillo = cellFormat(backgroundColor=color(1.0, 0.9, 0.6)) # Falta algún dato (vacío)
fmt_rojo = cellFormat(backgroundColor=color(0.9, 0.6, 0.6))     # Datos dispares/error
fmt_gris = cellFormat(backgroundColor=color(0.85, 0.85, 0.85)) # Serie no encontrada en SAP

formatos_a_aplicar = []
col_letra = chr(ord('A') + len(encabezados_h2) - 1) if len(encabezados_h2) <= 26 else 'Z'

# Encontrar columna Número de Serie en Hoja 2
idx_sn_h2 = next((i for i, h in enumerate(encabezados_h2) if 'SERIE' in h or 'SN' in h), None)

if idx_sn_h2 is None:
    raise Exception("No se encontró la columna de Número de Serie en la Hoja 2.")

# 5. Evaluación Fila por Fila
print("--- RESUMEN DE COMPARACIÓN ---")
for i, fila_raw in enumerate(datos_hoja2[1:], start=2):
    fila_h2 = [normalizar(c) for c in fila_raw]

    if len(fila_h2) <= idx_sn_h2:
        continue

    num_serie = fila_h2[idx_sn_h2]
    rango_fila = f"A{i}:{col_letra}{i}"

    if not num_serie:
        continue

    # Caso A: El Número de Serie NO existe en SAP
    if num_serie not in db_referencia:
        print(f"Fila {i} [SN: {num_serie}]: ❌ NO ENCONTRADO EN SAP -> Coloreando GRIS")
        formatos_a_aplicar.append((rango_fila, fmt_gris))
        continue

    # Caso B: El Número de Serie SÍ existe en SAP
    fila_ref = db_referencia[num_serie]
    diferencias = []
    faltantes = []

    for idx_col, col_nombre in enumerate(encabezados_h2):
        if 'SERIE' in col_nombre or 'SN' in col_nombre:
            continue

        val_h2 = fila_h2[idx_col] if idx_col < len(fila_h2) else ""

        # Buscar en SAP una columna con nombre similar
        col_ref_key = next((k for k in fila_ref.keys() if col_nombre in k or k in col_nombre), None)
        val_ref = fila_ref.get(col_ref_key, "") if col_ref_key else ""

        if val_h2 != val_ref:
            if val_h2 == "" or val_ref == "":
                faltantes.append(f"{col_nombre} (H2:'{val_h2}' vs SAP:'{val_ref}')")
            else:
                diferencias.append(f"{col_nombre} (H2:'{val_h2}' vs SAP:'{val_ref}')")

    # Clasificación por colores
    if diferencias:
        print(f"Fila {i} [SN: {num_serie}]: 🛑 ERROR EN -> {', '.join(diferencias)}")
        formatos_a_aplicar.append((rango_fila, fmt_rojo))
    elif faltantes:
        print(f"Fila {i} [SN: {num_serie}]: ⚠️ FALTA DATO EN -> {', '.join(faltantes)}")
        formatos_a_aplicar.append((rango_fila, fmt_amarillo))
    else:
        print(f"Fila {i} [SN: {num_serie}]: ✅ COINCIDENCIA EXACTA")
        formatos_a_aplicar.append((rango_fila, fmt_verde))

# 6. Aplicación de colores masiva
if formatos_a_aplicar:
    print("\nAplicando colores en Google Sheets...")
    format_cell_ranges(hoja2, formatos_a_aplicar)
    print("✨ ¡Proceso completado con éxito!")
else:
    print("\nNo se registraron filas para colorear.")
````

````python
import re
import pandas as pd
import gspread
from google.colab import auth
import google.auth

# ------------------------------------------------------------------------------
# 1. AUTENTICACIÓN Y CONEXIÓN
# ------------------------------------------------------------------------------
auth.authenticate_user()
creds, _ = google.auth.default()
gc = gspread.authorize(creds)

NOMBRE_ARCHIVO = "TV"
sh = gc.open(NOMBRE_ARCHIVO)

def normalizar(texto):
    if texto is None:
        return ""
    txt = str(texto).strip().upper()
    return re.sub(r'\s+', ' ', txt)

# ------------------------------------------------------------------------------
# 2. CARGAR DICCIONARIO DE REFERENCIA (Copia de TR y Copia de TB)
# ------------------------------------------------------------------------------
HOJAS_REFERENCIA = ['Copia de TR', 'Copia de TB']
db_referencia = {}

for nombre in HOJAS_REFERENCIA:
    try:
        wks_ref = sh.worksheet(nombre)
        datos = wks_ref.get_all_records()
        for fila in datos:
            # Buscar la llave del número de serie dinámicamente
            col_sn = next((k for k in fila.keys() if 'SERIE' in str(k).upper() or 'SN' in str(k).upper()), None)
            if col_sn:
                sn_val = normalizar(fila[col_sn])
                if sn_val:
                    # Guardar fila con claves normalizadas
                    db_referencia[sn_val] = {normalizar(k): str(v).strip() for k, v in fila.items()}
    except gspread.WorksheetNotFound:
        print(f"⚠️ No se encontró la hoja '{nombre}'.")

print(f"✅ Registros de referencia cargados desde Copias: {len(db_referencia)}")

# ------------------------------------------------------------------------------
# 3. LEER Y PROCESAR HOJA 2
# ------------------------------------------------------------------------------
wks_h2 = sh.worksheet('Hoja 2')
datos_h2_raw = wks_h2.get_all_values()

if not datos_h2_raw:
    raise Exception("La 'Hoja 2' está vacía.")

headers_h2 = [normalizar(h) for h in datos_h2_raw[0]]
idx_sn_h2 = next((i for i, h in enumerate(headers_h2) if 'SERIE' in h or 'SN' in h), None)

if idx_sn_h2 is None:
    raise Exception("No se encontró la columna de Número de Serie en Hoja 2.")

# Columnas estándar para la planilla de Coincidencias
columnas_salida = [
    'NUMERO DE SERIE',
    'ORIGEN',
    'MARCA (TI - TV)',
    'MODELO',
    'RELACION (TV)',
    'PRESTACIÓN (TV)',
    'CLASE (TV)',
    'AÑO FABRICACION',
    'Aislación (SF6/ACEITE/RESINA)'
]

matriz_coincidencias = [columnas_salida]
filas_noe = [datos_h2_raw[0]] # Encabezados originales para NOE

# Mapeo de campos estándar vs alias en los datos
def obtener_valor(dict_fila, alias_list):
    for alias in alias_list:
        match_key = next((k for k in dict_fila.keys() if alias in k), None)
        if match_key:
            return dict_fila[match_key]
    return ""

# Procesamiento de cada fila de Hoja 2
for fila_raw in datos_h2_raw[1:]:
    if len(fila_raw) <= idx_sn_h2:
        continue

    num_serie = normalizar(fila_raw[idx_sn_h2])
    if not num_serie:
        continue

    # Convertir la fila actual de Hoja 2 a diccionario
    dict_h2 = {headers_h2[i]: fila_raw[i].strip() if i < len(fila_raw) else "" for i in range(len(headers_h2))}

    if num_serie in db_referencia:
        dict_ref = db_referencia[num_serie]

        # Fila 1 del par: Datos de Copias (SAP)
        fila_copia = [
            num_serie,
            'COPIA SAP',
            obtener_valor(dict_ref, ['MARCA', 'FABRICANTE']),
            obtener_valor(dict_ref, ['MODELO', 'TIPO']),
            obtener_valor(dict_ref, ['RELACION']),
            obtener_valor(dict_ref, ['PRESTACIÓN', 'PRESTACION', 'CARGA']),
            obtener_valor(dict_ref, ['CLASE']),
            obtener_valor(dict_ref, ['AÑO', 'ANIO']),
            obtener_valor(dict_ref, ['AISLACIÓN', 'AISLACION', 'FLUIDO'])
        ]

        # Fila 2 del par: Datos de Hoja 2
        fila_hoja2 = [
            num_serie,
            'HOJA 2',
            obtener_valor(dict_h2, ['MARCA', 'FABRICANTE']),
            obtener_valor(dict_h2, ['MODELO', 'TIPO']),
            obtener_valor(dict_h2, ['RELACION']),
            obtener_valor(dict_h2, ['PRESTACIÓN', 'PRESTACION', 'CARGA']),
            obtener_valor(dict_h2, ['CLASE']),
            obtener_valor(dict_h2, ['AÑO', 'ANIO']),
            obtener_valor(dict_h2, ['AISLACIÓN', 'AISLACION', 'FLUIDO'])
        ]

        matriz_coincidencias.append(fila_copia)
        matriz_coincidencias.append(fila_hoja2)
    else:
        # No encontrado en Copias -> a planilla NOE
        filas_noe.append(fila_raw)

# ------------------------------------------------------------------------------
# 4. ESCRIBIR Y COMBINAR CELDAS EN HOJA 'Coincidencias'
# ------------------------------------------------------------------------------
try:
    wks_coinc = sh.worksheet('Coincidencias')
    wks_coinc.clear()
except gspread.WorksheetNotFound:
    wks_coinc = sh.add_worksheet(title='Coincidencias', rows=100, cols=15)

wks_coinc.update(range_name='A1', values=matriz_coincidencias)

# Petición masiva para unir las celdas de a dos en la columna 'NUMERO DE SERIE' (Columna A)
total_filas = len(matriz_coincidencias)
peticiones_merge = []

for r in range(1, total_filas, 2): # r inicia en fila index 1 (fila 2 de la planilla)
    peticiones_merge.append({
        "mergeCells": {
            "range": {
                "sheetId": wks_coinc.id,
                "startRowIndex": r,
                "endRowIndex": r + 2, # Combina de a 2 filas
                "startColumnIndex": 0,
                "endColumnIndex": 1
            },
            "mergeType": "MERGE_ALL"
        }
    })

# Centrar verticalmente la celda combinada del número de serie
peticiones_merge.append({
    "repeatCell": {
        "range": {
            "sheetId": wks_coinc.id,
            "startRowIndex": 1,
            "endRowIndex": total_filas,
            "startColumnIndex": 0,
            "endColumnIndex": 1
        },
        "cell": {
            "userEnteredFormat": {
                "verticalAlignment": "MIDDLE",
                "horizontalAlignment": "CENTER"
            }
        },
        "fields": "userEnteredFormat(verticalAlignment,horizontalAlignment)"
    }
})

if peticiones_merge:
    sh.batch_update({"requests": peticiones_merge})

print(f"✅ Hoja 'Coincidencias' generada con {(total_filas - 1) // 2} entradas pareadas.")

# ------------------------------------------------------------------------------
# 5. ESCRIBIR HOJA 'NOE'
# ------------------------------------------------------------------------------
try:
    wks_noe = sh.worksheet('NOE')
    wks_noe.clear()
except gspread.WorksheetNotFound:
    wks_noe = sh.add_worksheet(title='NOE', rows=100, cols=15)

wks_noe.update(range_name='A1', values=filas_noe)
print(f"✅ Hoja 'NOE' generada con {len(filas_noe) - 1} entradas sin coincidencia.")
````
